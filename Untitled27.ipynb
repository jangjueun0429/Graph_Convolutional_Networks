```python
import os
import torch
import numpy as np
import pandas as pd
import yfinance as yf
import joblib
from tqdm.auto import tqdm
from transformers import AutoTokenizer, AutoModelForSequenceClassification
from sklearn.preprocessing import RobustScaler
from sklearn.ensemble import IsolationForest

print("====== 🚀 [1단계] 환경 세팅 및 디바이스 확인 ======")
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"🔹 현재 사용 중인 하드웨어 가속기: {device}")
if device.type != 'cuda':
    print("⚠ 경고: GPU 가속기가 꺼져 있습니다. 상단 메뉴 [런타임] -> [런타임 유형 변경]에서 T4 GPU를 켜시면 속도가 10배 이상 빨라집니다!")

# =====================================================================
# [Train 파트] 2. 2017 ~ 2022년 데이터 통합 및 모델 학습
# =====================================================================
print("\n====== 🚀 [2단계: Train] 2017~2022년 데이터 통합 및 학습 시작 ======")

# 주은님이 업로드한 파일 이름 리스트와 정확히 매칭 (2017~2022)
train_files = [f'/content/news_train_{year}.xlsx' for year in range(2017, 2023)]

print("🔹 훈련 뉴스 엑셀 파일 로드 및 병합 중...")
train_dfs = []
for f in train_files:
    if os.path.exists(f):
        train_dfs.append(pd.read_excel(f))
    else:
        print(f"⚠ 경고: {f} 파일이 업로드되지 않았습니다. 건너넙니다.")

if not train_dfs:
    raise FileNotFoundError("🔴 Train 파일을 단 하나도 찾을 수 없습니다! 왼쪽 파일 창의 이름을 확인해 주셈.")

train_news_df = pd.concat(train_dfs, axis=0, ignore_index=True)
train_news_df['일자'] = pd.to_datetime(train_news_df['일자'], errors='coerce')
train_news_df = train_news_df.dropna(subset=['일자', '제목']).sort_values(by='일자').reset_index(drop=True)
print(f"✅ Train 뉴스 통합 완료! 총 건수: {train_news_df.shape[0]}건")

# KR-FinBERT-SC 감성 분석 모델 및 토크나이저 로드
print("\n🔹 서울대 KR-FinBERT-SC 모델 로딩 중...")
tokenizer = AutoTokenizer.from_pretrained("snunlp/KR-FinBert-SC")
model = AutoModelForSequenceClassification.from_pretrained("snunlp/KR-FinBert-SC").to(device)
model.eval()

def convert_scores_batch(titles, batch_size=64):
    all_scores = []
    cleaned_titles = [str(title) for title in titles]
    for i in tqdm(range(0, len(cleaned_titles), batch_size), desc="BERT 감성 분석 진행 중"):
        batch_titles = cleaned_titles[i:i+batch_size]
        inputs = tokenizer(batch_titles, padding=True, truncation=True, return_tensors="pt").to(device)
        with torch.no_grad():
            outputs = model(**inputs)
            probs = torch.softmax(outputs.logits, dim=-1)
        pred_labels = torch.argmax(probs, dim=-1).cpu().numpy()
        pred_scores = torch.max(probs, dim=-1).values.cpu().numpy()
        for label, score in zip(pred_labels, pred_scores):
            if label == 0: all_scores.append(-score) # 부정
            elif label == 1: all_scores.append(0.0)  # 중립
            else: all_scores.append(score)           # 긍정
    return all_scores

print("🔹 훈련 데이터 뉴스 감성 정량화 시작 (시간이 몇 분 소요될 수 있음)...")
train_news_df['Sentiment'] = convert_scores_batch(train_news_df['제목'].tolist())
train_daily_sentiment = train_news_df.groupby('일자')['Sentiment'].mean().to_frame()

# 야후 파이낸스 과거 데이터 수집
print("\n🔹 과거 금융 지표 다운로드 중...")
tickers = {"KOSDAQ": "^KQ11", "VIX": "^VIX", "NASDAQ": "^IXIC", "WTI": "CL=F"}
raw_data = yf.download(list(tickers.values()), start="2017-01-01", end="2022-12-31", group_by='ticker')

financial_series = {}
for name, ticker in tickers.items():
    financial_series[name] = raw_data[ticker]['Adj Close'] if 'Adj Close' in raw_data[ticker].columns else raw_data[ticker]['Close']

train_financial_df = pd.DataFrame(financial_series).dropna()
train_log_df = np.log(train_financial_df / train_financial_df.shift(1)).dropna()

train_log_df.index = pd.to_datetime(train_log_df.index)
train_daily_sentiment.index = pd.to_datetime(train_daily_sentiment.index)
train_final_df = train_log_df.join(train_daily_sentiment, how='left').fillna(0)

# 스케일러 및 국면 모델 학습 후 파일로 박제 (.pkl)
feature_cols = ['KOSDAQ', 'VIX', 'NASDAQ', 'WTI']
train_scaled_df = train_final_df.copy()

scaler = RobustScaler()
train_scaled_df[feature_cols] = scaler.fit_transform(train_final_df[feature_cols])
joblib.dump(scaler, '/content/robust_scaler.pkl') # 기준 저장 1

iso_forest = IsolationForest(contamination=0.1509, random_state=42)
train_regime_preds = iso_forest.fit_predict(train_scaled_df[feature_cols])
joblib.dump(iso_forest, '/content/isolation_forest_model.pkl') # 기준 저장 2

print("✅ 과거 기준 가중치 파일(.pkl) 2개 생성 및 저장 완료!")


# =====================================================================
# [OOT 파트] 3. 2023 ~ 2026년 검증 데이터 구축 및 과거 기준 적용
# =====================================================================
print("\n====== 🚀 [3단계: OOT Test] 2023~2026년 현재 데이터 구축 시작 ======")

# 검증용 파일 목록 확인 (2023~2026)
oot_files = [f'/content/news_oot_{year}.xlsx' for year in range(2023, 2027)]

print("🔹 OOT 검증 뉴스 엑셀 파일 로드 및 병합 중...")
oot_dfs = []
for f in oot_files:
    if os.path.exists(f):
        oot_dfs.append(pd.read_excel(f))
    else:
        print(f"⚠ 경고: {f} 파일이 업로드되지 않았습니다. 건너넙니다.")

if not oot_dfs:
    raise FileNotFoundError("🔴 OOT 검증 파일을 단 하나도 찾을 수 없습니다! 파일명을 다시 확인해 주셈.")

oot_news_df = pd.concat(oot_dfs, axis=0, ignore_index=True)
oot_news_df['일자'] = pd.to_datetime(oot_news_df['일자'], errors='coerce')
oot_news_df = oot_news_df.dropna(subset=['일자', '제목']).sort_values(by='일자').reset_index(drop=True)
print(f"✅ OOT 뉴스 통합 완료! 총 건수: {oot_news_df.shape[0]}건")

print("\n🔹 검증 데이터 뉴스 감성 정량화 시작...")
oot_news_df['Sentiment'] = convert_scores_batch(oot_news_df['제목'].tolist())
oot_daily_sentiment = oot_news_df.groupby('일자')['Sentiment'].mean().to_frame()

# 야후 파이낸스 최신 데이터 수집 (2023년부터 2026년 오늘 날짜까지)
print("\n🔹 최신 금융 지표 오늘 날짜까지 다운로드 중...")
raw_data_oot = yf.download(list(tickers.values()), start="2023-01-01", end="2026-05-29", group_by='ticker')

financial_series_oot = {}
for name, ticker in tickers.items():
    financial_series_oot[name] = raw_data_oot[ticker]['Adj Close'] if 'Adj Close' in raw_data_oot[ticker].columns else raw_data_oot[ticker]['Close']

oot_financial_df = pd.DataFrame(financial_series_oot).dropna()
oot_log_df = np.log(oot_financial_df / oot_financial_df.shift(1)).dropna()

oot_log_df.index = pd.to_datetime(oot_log_df.index)
oot_daily_sentiment.index = pd.to_datetime(oot_daily_sentiment.index)
oot_final_df = oot_log_df.join(oot_daily_sentiment, how='left').fillna(0)

# 2단계에서 박제했던 과거 가중치 (.pkl) 실시간 로드 및 적용
scaler_loaded = joblib.load('/content/robust_scaler.pkl')
iso_forest_loaded = joblib.load('/content/isolation_forest_model.pkl')

oot_scaled_df = oot_final_df.copy()
oot_scaled_df[feature_cols] = scaler_loaded.transform(oot_final_df[feature_cols])

# 과거 훈련 데이터 기준으로 최신 위기 국면(Regime) 판정
oot_regime_preds = iso_forest_loaded.predict(oot_scaled_df[feature_cols])
oot_final_df['Regime'] = np.where(oot_regime_preds == -1, 1, 0)
oot_scaled_df['Regime'] = oot_final_df['Regime']

print("\n====== 📊 [최종 통일 완료] 2023~2026 OOT 데이터셋 결과 ======")
print(f"총 검증 영업일수: {oot_scaled_df.shape[0]}일")
print(oot_scaled_df[['KOSDAQ', 'VIX', 'NASDAQ', 'WTI', 'Sentiment', 'Regime']].head(10))
```

```python
import numpy as np
import torch
from torch.utils.data import DataLoader, TensorDataset

print("====== 🚀 [3단계] SimpleTFT 입력용 3차원 텐서 변환 및 데이터로더 제작 시작 ======")

# =====================================================================
# 1. 윈도우 크기(과거 타임스텝) 설정
# =====================================================================
# 지난 21일간의 마켓 흐름을 보고 다음 날을 예측 (학습 때 세팅과 동일)
window_size = 21

# 모델 입력으로 들어갈 5개 특성 변수 선택
# (스케일링이 완료된 oot_scaled_df를 사용함)
feature_cols = ['KOSDAQ', 'VIX', 'NASDAQ', 'WTI', 'Sentiment']

X_oot_raw = oot_scaled_df[feature_cols].values
y_oot_raw = oot_scaled_df['Regime'].values

X_oot_list = []
y_oot_list = []

# =====================================================================
# 2. 윈도우 슬라이싱 (Sliding Window) 수행
# =====================================================================
# 21일씩 데이터를 묶고, 바로 다음날(t + window_size)의 Regime을 타겟으로 잡음
for i in range(len(X_oot_raw) - window_size):
    X_oot_list.append(X_oot_raw[i : i + window_size])
    y_oot_list.append(y_oot_raw[i + window_size])

# 넘파이 배열로 변환
X_oot_np = np.array(X_oot_list, dtype=np.float32)
y_oot_np = np.array(y_oot_list, dtype=np.float32)

print(f"✅ 윈도우 슬라이싱 완료!")
print(f"   - 입력 데이터(X) 모양: {X_oot_np.shape} ➡️ (샘플 수, 윈도우 크기, 변수 개수)")
print(f"   - 정답 데이터(y) 모양: {y_oot_np.shape} ➡️ (샘플 수,)")

# =====================================================================
# 3. 파이토치 텐서(Tensor) 변환 및 데이터로더(DataLoader) 구축
# =====================================================================
# 넘파이 배열을 파이토치 모델이 인식할 수 있는 텐서로 변환
X_oot_tensor = torch.tensor(X_np, dtype=torch.float32) if 'X_np' in locals() else torch.tensor(X_oot_np, dtype=torch.float32)
y_oot_tensor = torch.tensor(y_oot_np, dtype=torch.float32)

# SimpleTFT 모델에 배치(Batch) 단위로 안전하게 밀어넣기 위한 데이터로더 객체 생성
# 검증 단계이므로 굳이 셔플(shuffle)하지 않고 순서대로 순방향 검증 진행
oot_dataset = TensorDataset(X_oot_tensor, y_oot_tensor)
oot_loader = DataLoader(oot_dataset, batch_size=64, shuffle=False)

print("\n====== 📊 최종 SimpleTFT 인풋용 텐서 구축 결과 ======")
print(f"🔹 총 슬라이싱된 검증 시퀀스 개수: {X_oot_tensor.shape[0]}개")
print(f"🔹 각 시퀀스별 타임스텝(Lookback Period): {X_oot_tensor.shape[1]}일")
print(f"🔹 모델이 참조할 독립변수(Features) 개수: {X_oot_tensor.shape[2]}개")
print("✅ SimpleTFT 모델에게 모의고사 문제를 출제할 준비가 끝났습니다!")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
import numpy as np
import os
import joblib

print("====== 🚀 [4단계-1] SimpleTFT 모델 구조 선언 및 안전 학습 시작 ======")

# =====================================================================
# 0. [안전 장치] 데이터프레임 확인 및 변수 복구
# =====================================================================
if 'train_scaled_df' not in locals():
    if 'train_final_df' in locals():
        train_scaled_df = train_final_df.copy()
    else:
        raise NameError("🔴 Train 전처리 데이터프레임이 메모리에 없습니다. 위쪽의 통합 전처리 셀을 다시 한 번 실행해 주세요!")

window_size = 21
feature_cols = ['KOSDAQ', 'VIX', 'NASDAQ', 'WTI', 'Sentiment']
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

# =====================================================================
# 0-1. [★에러 해결★] Isolation Forest 학습 규격에 맞게 4개 컬럼만 전달
# =====================================================================
# Isolation Forest는 'Sentiment'를 제외한 4개 정형 지표로만 국면을 판정했습니다.
iso_features = ['KOSDAQ', 'VIX', 'NASDAQ', 'WTI']

try:
    if 'iso_forest' in locals():
        # 학습 규격에 맞춰 4개 변수만 슬라이싱하여 예측
        train_regime_preds = iso_forest.predict(train_scaled_df[iso_features])
    else:
        iso_forest_loaded = joblib.load('/content/isolation_forest_model.pkl')
        train_regime_preds = iso_forest_loaded.predict(train_scaled_df[iso_features])

    # 정상(1) -> 0, 위기(-1) -> 1 로 매핑하여 Regime 컬럼 생성
    train_scaled_df['Regime'] = np.where(train_regime_preds == -1, 1, 0)
    print("✅ Train 데이터프레임 내 'Regime' 정답 컬럼 매핑 성공!")
except Exception as e:
    print(f"❌ Isolation Forest 기반 라벨 복구 실패: {e}")
    raise

# =====================================================================
# 1. 주은님의 연구 목적에 특화된 SimpleTFT 아키텍처 정의
# =====================================================================
class SimpleTFT(nn.Module):
    def __init__(self, input_dim, hidden_dim, output_dim=1, num_layers=1):
        super(SimpleTFT, self).__init__()
        # Variable Selection Network (VSN) 역할: 5개 변수 중 무엇이 중요한지 동적 가중치 산출
        self.vsn = nn.Linear(input_dim, input_dim)
        # 거시경제 시계열의 장단기 흐름을 추적하는 핵심 LSTM 레이어
        self.lstm = nn.LSTM(input_dim, hidden_dim, num_layers, batch_first=True)
        # 최종적으로 오늘이 '위기 국면(1)'일 확률을 0.0 ~ 1.0 사이로 출력하는 Dense + Sigmoid Layer
        self.fc = nn.Linear(hidden_dim, output_dim)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        # 1. 변수 중요도 동적 부여
        vsn_weights = torch.softmax(self.vsn(x), dim=-1)
        x = x * vsn_weights

        # 2. 시계열 컨텍스트 추출
        lstm_out, (hn, cn) = self.lstm(x)
        last_hidden = lstm_out[:, -1, :] # 21일의 타임스텝 중 가장 마지막 날의 요약 정보 추출

        # 3. 국면 확률 출력
        out = self.fc(last_hidden)
        return self.sigmoid(out)

# =====================================================================
# 2. 훈련 데이터(2017~2022) 슬라이싱 빌드 (SimpleTFT 입력용은 5개 변수 통째로 사용)
# =====================================================================
X_train_raw = train_scaled_df[feature_cols].values
y_train_raw = train_scaled_df['Regime'].values

X_train_list, y_train_list = [], []
for i in range(len(X_train_raw) - window_size):
    X_train_list.append(X_train_raw[i : i + window_size])
    y_train_list.append(y_train_raw[i + window_size])

X_train_tensor = torch.tensor(np.array(X_train_list), dtype=torch.float32)
y_train_tensor = torch.tensor(np.array(y_train_list), dtype=torch.float32)

train_dataset = TensorDataset(X_train_tensor, y_train_tensor)
train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)

# =====================================================================
# 3. 모델 컴파일 및 T4 GPU 가속 기반 훈련 실행
# =====================================================================
train_model = SimpleTFT(input_dim=5, hidden_dim=64).to(device)
criterion = nn.BCELoss() # 이진 분류를 위한 Binary Cross Entropy Loss
optimizer = optim.Adam(train_model.parameters(), lr=0.001)

train_model.train()
for epoch in range(20):
    epoch_loss = 0
    for batch_x, batch_y in train_loader:
        batch_x, batch_y = batch_x.to(device), batch_y.to(device)
        optimizer.zero_grad()
        outputs = train_model(batch_x).squeeze()
        loss = criterion(outputs, batch_y)
        loss.backward()
        optimizer.step()
        epoch_loss += loss.item()
    if (epoch+1) % 5 == 0:
        print(f"   Epoch [{epoch+1}/20] - Loss: {epoch_loss/len(train_loader):.4f}")

# 완성된 가중치를 로컬 디렉토리에 백업 저장
torch.save(train_model.state_dict(), '/content/simple_tft_model.pth')
print("✅ [성공] SimpleTFT 모델 즉석 학습 및 /content/simple_tft_model.pth 저장 완료!")
```

```python
print("====== 🚀 [4단계-2] 저장된 SimpleTFT 모델 로드 및 OOT 실전 예측 시작 ======")

model_path = '/content/simple_tft_model.pth'
model = SimpleTFT(input_dim=5, hidden_dim=64).to(device)

# 백업된 가중치 로드
model.load_state_dict(torch.load(model_path, map_location=device))
model.eval()
print("✅ 가중치 복원 완료! OOT 추론(Inference) 단계로 진입합니다.")

oot_preds = []
oot_targets = []

# 순방향 시계열 예측 수행 (그레디언트 추적 차단하여 메모리 세이브)
with torch.no_grad():
    for batch_x, batch_y in oot_loader:
        batch_x = batch_x.to(device)
        pred = model(batch_x)
        oot_preds.extend(pred.cpu().numpy().flatten())
        oot_targets.extend(batch_y.numpy())

# 시각화 데이터와 맵핑하기 위해 윈도우 크기(21일) 이후의 날짜 인덱스로 정렬
oot_result_df = oot_scaled_df.iloc[window_size:].copy()
oot_result_df['Pred_Probability'] = oot_preds # 모델이 계산한 확률값

# 임계치(Threshold) 0.5 기준으로 최종 위기 국면(1)과 평시(0)로 이진 분류
oot_result_df['Pred_Regime'] = np.where(oot_result_df['Pred_Probability'] >= 0.5, 1, 0)

print("\n====== 📊 SimpleTFT 모델 2023~2026 예측 완료 ======")
print(f"🔹 총 예측 영업일수: {len(oot_result_df)}일")
print(oot_result_df[['Pred_Probability', 'Pred_Regime', 'Regime']].head(10))
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("====== 🚀 [7단계] 최종 연구가설 01: DeltaLag 기반 동적 시차 수축 효과 검증 ======")

# =====================================================================
# 1. 국면별(Normal vs Risk) 데이터 분리 및 시차별 상관 강도 연산
# =====================================================================
# 타겟 변수: 코스닥 내일 수익률 (t+1)
# 원인 변수: 과거 t-1, t-2, ... t-5일 전의 매크로 및 심리 지표

features_to_analyze = ['VIX', 'NASDAQ', 'WTI', 'Sentiment']
max_lag = 5 # D-1부터 D-5까지의 시차 분석

# 분석용 임시 데이터프레임 카피
lag_analysis_df = oot_result_df.copy()
# 내일의 코스닥 수익률 생성 (타겟)
lag_analysis_df['KOSDAQ_Target'] = lag_analysis_df['KOSDAQ'].shift(-1)
lag_analysis_df = lag_analysis_df.dropna()

def calculate_lag_dependence(df_subset):
    """각 시차(Lag)별로 변수들이 타겟에 미치는 평균 충격 강도를 계산"""
    lag_matrix = {var: [] for var in features_to_analyze}

    for var in features_to_analyze:
        for lag in range(1, max_lag + 1):
            # 과거 시차 데이터와 미래 타겟 간의 상호 상관관계(Cross-Correlation)의 절대값 추출
            correlation = df_subset[var].shift(lag).corr(df_subset['KOSDAQ_Target'])
            lag_matrix[var].append(abs(correlation))

    return pd.DataFrame(lag_matrix, index=[f'D-{i}' for i in range(1, max_lag + 1)])

# 2. SimpleTFT가 판정한 국면별로 데이터 쪼개기
normal_subset = lag_analysis_df[lag_analysis_df['Pred_Regime'] == 0]
risk_subset = lag_analysis_df[lag_analysis_df['Pred_Regime'] == 1]

normal_lag_df = calculate_lag_dependence(normal_subset).fillna(0)
risk_lag_df = calculate_lag_dependence(risk_subset).fillna(0)

# =====================================================================
# 3. 학술 논문용 DeltaLag 시차 수축 그래프 시각화
# =====================================================================
fig, axes = plt.subplots(1, 2, figsize=(16, 6), sharey=True)
lag_ticks = [f'D-{i}' for i in range(1, max_lag + 1)]

# (1) Normal Regime 그래프
for var in features_to_analyze:
    axes[0].plot(lag_ticks, normal_lag_df[var], marker='o', linewidth=2, label=var)
axes[0].set_title('Information Transmit Lag: Normal Regime (평상시)', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Time Lag (Days Before)', fontsize=11)
axes[0].set_ylabel('Cross-Attention Shock Intensity', fontsize=11)
axes[0].grid(True, linestyle='--', alpha=0.5)
axes[0].legend()

# (2) Risk Regime 그래프
for var in features_to_analyze:
    axes[1].plot(lag_ticks, risk_lag_df[var], marker='s', linewidth=2, linestyle='--', label=var)
axes[1].set_title('Information Transmit Lag: Risk Regime (위기상황)', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Time Lag (Days Before)', fontsize=11)
axes[1].grid(True, linestyle='--', alpha=0.5)
axes[1].legend()

plt.suptitle('DeltaLag Analysis: Evidence of Information Lag Contraction', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

# =====================================================================
# 4. 최적 선행 시차(Optimal Lag) 도출 및 검증 출력
# =====================================================================
print("\n==================== 📊 DeltaLag 시차 수축 검증 표 ====================")
print("[평상시 (Normal) 각 변수별 피크 충격 시차]")
for var in features_to_analyze:
    optimal_lag_normal = normal_lag_df[var].idxmax()
    print(f"  🔹 {var:text<10} : {optimal_lag_normal} (최대 강도: {normal_lag_df[var].max():.4f})")

print("\n[위기상황 (Risk) 각 변수별 피크 충격 시차]")
for var in features_to_analyze:
    optimal_lag_risk = risk_lag_df[var].idxmax()
    print(f"  🚨 {var:text<10} : {optimal_lag_risk} (최대 강도: {risk_lag_df[var].max():.4f})")
print("=======================================================================")
```
